```python

url = "http://pythonscraping.com/pages/page1.html"

try:
    html = urlopen(url)
    soup = BeautifulSoup(html.read(), "html.parser")

    # Ambil elemen judul <h1> dan teks di dalamnya
    print("Judul H1:", soup.h1.get_text())
    print("\nIsi Konten HTML:")
    print(soup.prettify())
except Exception as e:
    print("Terjadi kesalahan:", e)
```

```python
# Fungsi untuk mengambil judul halaman dengan penanganan HTTP/URL Error & AttributeError
def getTitle(url):
    try:
        html = urlopen(url)
    except (HTTPError, URLError) as e:
        print(f"Peringatan Koneksi/Server: {e}")
        return None

    try:
        bs = BeautifulSoup(html.read(), 'html.parser')
        title = bs.body.h1
    except AttributeError as e:
        print(f"Peringatan Tag HTML: Tag tidak ditemukan ({e})")
        return None

    return title

# Pengujian fungsi
target_url = "http://www.pythonscraping.com/pages/page1.html"
title = getTitle(target_url)

if title is None:
    print('Title tidak dapat ditemukan!')
else:
    print("Hasil eksekusi getTitle():", title.get_text())
```

```python
# Scraping daftar judul dari halaman web dan menyimpannya ke CSV
url = "https://chomsky.info/audionvideo/"

# Menggunakan User-Agent agar request tidak diblokir server
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}

try:
    response = requests.get(url, headers=headers)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    # Mencari seluruh elemen daftar <li>
    titles = soup.find_all("li")

    data = []
    for item in titles:
        text = item.get_text(strip=True)
        if text:  # Pastikan teks tidak kosong
            data.append(text)

    # Menyimpan ke dalam DataFrame Pandas
    df = pd.DataFrame(data, columns=["title"])

    # Export ke file CSV
    output_filename = "chomsky.csv"
    df.to_csv(output_filename, index=False)

    print(f"Berhasil mengikis {len(df)} data dan menyimpan ke '{output_filename}'!")

    # Menampilkan 5 data teratas di Google Colab
    display(df.head())

except Exception as e:
    print(f"Gagal melakukan scraping: {e}")
```
